
# 📓 Creating Azure Data Factory Resources for CI/CD

In this note, we will create the required **Azure Data Factory (ADF)** resources for implementing CI/CD.  
We will **not reuse** the Data Factory from the COVID project — instead, we will create new ones with naming standards aligned to DevOps practices.

---

## 🎯 Why Create New Data Factories?

We want a **clean setup** to implement CI/CD from scratch, so that we can:

✅ Apply proper naming conventions  
✅ Configure Git integration from the beginning  
✅ Observe CI/CD effects across environments  

---

## 🏗️ Required Data Factories for CI/CD

A standard Azure DevOps project typically requires multiple ADF instances across environments:

| Environment | Purpose | Git Enabled? | Notes |
|-------------|-----------|----------------|--------|
| **Dev** | Development & CI | ✅ Yes | Developers commit & collaborate |
| **Test** | Testing (Integration/UAT) | ❌ No | Code deployed via CD pipeline |
| **Prod** | Production execution | ❌ No | Final deployment stage |

> In enterprise environments, there may be additional environments (e.g., QA, UAT, Pre‑Prod).  
For simplicity, we will use **Dev → Test → Prod**.

---

## 🔖 Naming Standards

To maintain consistency and enable automation in pipelines, we follow a naming pattern:

```
<environment>-ci-cd-demo-<resource>
```

### 📍 Resource Naming Table

| Resource | Dev Name | Test Name | Prod Name |
|-----------|-------------|-------------|--------------|
| Resource Group | `dev-ci-cd-demo-rg` | `test-ci-cd-demo-rg` | `prod-ci-cd-demo-rg` |
| Data Factory | `dev-ci-cd-demo-adf` | `test-ci-cd-demo-adf` | `prod-ci-cd-demo-adf` |

💡 Only the **Dev ADF** will be connected to Git — Test & Prod ADFs receive code via deployment pipelines.

---

## ⚙️ Steps to Create the Data Factories

1. Sign into **Azure Portal** → https://portal.azure.com  
2. Navigate to **Create a Resource**
3. Search for **Data Factory**
4. Create the Dev Data Factory first:

| Field | Value |
|--------|---------|
| Subscription | Your Subscription |
| Resource Group | `dev-ci-cd-demo-rg` |
| Data Factory Name | `dev-ci-cd-demo-adf` |
| Region | Closest to you (e.g., UK South) |
| Version | V2 |
| Git Configuration | Configure later |

5. Repeat the same steps to create Test & Prod Data Factories – without Git config.

---

## ✅ Final Check

After creation, you should see **three** Data Factories in Azure Portal:

| Data Factory | Resource Group |
|----------------|--------------------|
| dev-ci-cd-demo-adf | dev-ci-cd-demo-rg |
| test-ci-cd-demo-adf | test-ci-cd-demo-rg |
| prod-ci-cd-demo-adf | prod-ci-cd-demo-rg |
