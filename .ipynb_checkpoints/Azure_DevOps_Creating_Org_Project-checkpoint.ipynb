# **Creating Azure DevOps Organization & Project**

In the previous lesson, we understood what Azure DevOps is.  
In this lesson, we will walk through:

✅ Signing into Azure DevOps  
✅ Creating an Azure DevOps Organization  
✅ Linking to the correct Azure Active Directory (AAD)  
✅ Creating a Project inside the Organization  
✅ Quick tour of Azure DevOps services  
✅ Enabling free parallelism for CI/CD pipelines  

---


## 🔐 **Step 1: Sign in to Azure DevOps**

1. Open your browser and go to:  
   👉 https://dev.azure.com  
2. Click **Sign in to Azure DevOps**  
3. Select the same Microsoft account used for Azure Portal  
4. Enter password → **Sign In**  
5. Click **Continue**  

> ⚠️ Do **NOT** continue on the next screen yet — We must choose the correct directory first.


## 🧭 **Step 2: Select the Correct Azure Active Directory (AAD)**

Azure DevOps needs to be linked to the same directory where your Azure resources (like ADF) are created.

1. Click **Switch Directory**  
2. Identify the correct directory:
   - Go to https://portal.azure.com
   - On the top-right corner, click your profile → Check AAD name
3. Select the same directory in Azure DevOps  
4. Click **Continue**  

> Most users will see **Default Directory**.  
> For corporate environments, ensure you pick the correct organization-specific directory.


## 🏢 **Step 3: Create Azure DevOps Organization**

1. Select your country/region  
2. Enter a unique **Organization Name**  
   - Example: `john-devops-labs`
   - If name not available → append numbers/initials  
3. Enter CAPTCHA characters  
4. Click **Continue**  

Once created, you will land inside your Azure DevOps organization.

To view org details:  
`Organization Settings → Overview`  


## 💳 **Step 4: Understand Free Azure DevOps Licensing**

Azure DevOps offers a free tier suitable for learning and personal projects.

| Feature | Free Allowance |
|--------|----------------|
| Users | 5 users free |
| CI/CD (Microsoft hosted agents) | 1,800 mins/month |
| Self-hosted CI/CD | Unlimited |
| Artifacts Storage | 2 GB |

> This is more than sufficient for ADF + CICD learning.


## ⚙️ **Enable Free Parallelism (Optional but Recommended)**

Microsoft requires a request to enable **free parallel jobs**.

🔗 Form link:  
https://aka.ms/azpipelines-parallelism-request  

Fill the form:

| Field | What to Enter |
|-------|----------------|
| Name | Your name |
| Email | Your login email |
| DevOps Organization URL | Copy from Azure DevOps → Organization Overview |
| Project Type | Select **Private** |

Approval usually takes **< 48 hours**.


## 📁 **Step 5: Create a New Azure DevOps Project**

Two ways to create a project:

### ✅ Option 1: From Organization Home  
- Click **Azure DevOps** logo (top-left)  
- Click **New Project**  

### ✅ Option 2: From Projects Page  
- Click **Projects**  
- Click **New Project**  

### 📍 Provide the following:

| Field | Value |
|-------|--------|
| Project Name | Example: `ADF-CICD` |
| Visibility | **Private** |
| Version Control | **Git** (recommended) |
| Work Item Process | Basic / Agile / Scrum / CMMI |

> For learning: Choose **Basic**

Click **Create Project** ✅  


## 🚀 **Azure DevOps Project Layout (Quick Tour)**

Inside your project, you will see 5 core services:

| Service | Purpose |
|---------|----------|
| **Boards** | Requirements, backlog, sprints, Kanban |
| **Repos** | Git code repositories |
| **Pipelines** | Build + Release CICD |
| **Test Plans** | Manual testing & QA |
| **Artifacts** | Package management (like NPM, NuGet) |

For ADF CICD, we will use:

✔️ **Repos** – Store ADF code  
✔️ **Pipelines** – Build & Release CICD  

We will **not** use Boards, Test Plans, Artifacts for this course.
