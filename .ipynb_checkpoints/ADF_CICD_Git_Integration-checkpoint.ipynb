# Configuring Git Integration with Azure Data Factory
---
### 📘 Overview
This notebook provides detailed notes on how to configure a Git repository for Azure Data Factory (ADF) development. It also explains key ADF Git concepts such as collaboration branches, feature branches, and the `adf_publish` branch.

## 🔑 Key Git Concepts for ADF
Before configuring Git integration, it’s important to understand a few ADF‑specific Git terminologies:

### ✔ Recommended Branching Strategy
Microsoft recommends the following branching approach for ADF:

| Branch | Purpose |
|--------|-----------|
| **Main/Master (Collaboration Branch)** | The only branch from which code can be **Published** to ADF. No direct development allowed. |
| **Feature Branches** | Created from Main for development work. Merged via Pull Request (PR). |
| **adf_publish** | Auto‑generated branch after Publish. Contains ARM templates for deployment. |

🔹 **Developers must never develop directly on main.**

Publishing from main triggers creation/update of the `adf_publish` branch containing ARM templates required for deployments.

## 🧰 Step 1: Create a Git Repository in Azure DevOps
You can either:
- Create a repo in Azure DevOps first **(recommended)**, then link ADF to it, or
- Let ADF create the repo for you

Here is the recommended approach:

1. Go to Azure DevOps > Select your Project
2. Navigate to **Repos**
3. Click **New Repository**
4. Name the repository same as the ADF instance for easy mapping

> ✅ Recommended Naming: Use the same name as the ADF instance (e.g., `dev-ci-cd-demo-adf`)

## 🔧 Step 2: Configure Branch Policies
To enforce proper CI/CD workflows:

1. Go to **Repos → Branches**
2. Select the `main` branch
3. Click **... → Branch Policies**
4. Enable:
   - ✅ Minimum number of reviewers (e.g., 2 — but use 1 for learning/demo)
   - ✅ Require pull request to merge

📍 In real projects: *Do NOT allow requestors to approve their own PRs*

> This ensures quality code review and prevents direct commits to main.

## 🔗 Step 3: Configure Git in Azure Data Factory
1. Open **ADF Studio** for the Dev Data Factory
2. Navigate to one of the following:
   - Home screen → **Set up Code Repository**
   - Author tab → Git dropdown → **Set up Code Repository**
   - Manage tab → **Git Configuration → Configure**

### Fill in Configuration:
| Field | Value |
|-------|--------|
| Repository Type | Azure DevOps Git |
| Azure AD Tenant | Default Directory (or the one used for ADF resources) |
| DevOps Organization | Select your organization |
| Project | Select the DevOps project created |
| Repository | Select the repo created (e.g., `dev-ci-cd-demo-adf`) |
| Collaboration Branch | `main` or `master` |
| Publish Branch | `adf_publish` (auto‑selected) |

📍 If moving existing pipelines to Git, enable *Import Existing Resources*. For a clean start, leave unchecked.

## 🧪 After Configuration
Once the setup is complete:

- ADF switches to **Git mode** instead of Live mode
- You will see branch name at the top bar of ADF Studio
- You must now **create a feature branch** before making changes

⚠️ If you try to work on main branch, ADF will show an error because branch policies prevent direct edits.

## 🏁 Summary
You have now:
✅ Created a Git Repo in Azure DevOps
✅ Applied branch policies to protect the main branch
✅ Integrated ADF with Azure DevOps Git

This sets the foundation for CI (Continuous Integration) before enabling automated deployments (CD).
