#### Azure Components
- app service : it is part of compute domain
- basically there are 3 kinds of services in the compute domain
    - Virtual Machine: Azure VM is just like a new machine with a freshly installed operating system. (need this much RAM, Storage, configuration) and azure will create a VM for you and it will give you IP address and password for that machine. we have to do everything manually
    - App service: it is an automated version of virtual machine. you simply upload your code and it deploys the application. you just need to specify language you want, what kind of configuration of servers. here everything is Automatically
    - Functions: it is very new technology. it is called server less architecture(you don't have to specify the configuration). you just upload your code and it executes for you but it will deploy any application for you

- blob storage: it is basically the file system where you can store any kind of file. a place where storing the image files
     - blob storage: it is an all in one storage which can store any type of file
     - queue storage: it is used for queueing tasks which can read sequentially and can thus be used for Synchronization
     - backup: you can store all your infrequently accessed data at a minimum price
     
- MySQL for Azure is a MySQL Implementation on the Azure infrastructure 
- Auto Scaling and Load Balancing:  Auto Scale is used to automatically scale up and scale down according to needs, load balancer on the other hand is used to equally distributing the incoming data to the number of servers running currently
