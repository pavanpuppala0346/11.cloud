#### 🚀 <b>115.Introduction to CI/CD</b>

**Continuous Integration (CI)** and **Continuous Delivery (CD)** are key **DevOps practices** used to:
- Deliver software faster to the market  
- Increase confidence and quality in deployments

#### 🤝 Introduction to DevOps

Traditionally, organizations had **two separate teams**:

| Team | Responsibility |
|------|----------------|
| Development | Build and test software |
| IT Operations | Deploy and run applications in production |

#### ❗ Issues with Siloed Teams
- Poor communication and coordination
- Conflicting priorities across teams
- Limited automation and delays
- Slower time to market and poor product quality

#### ✅ What is DevOps?
DevOps was introduced around **2007** to:
- Bring Development and Operations together
- Work collaboratively throughout the software lifecycle
- Improve **speed, quality, and automation** of software delivery


#### 🧠 Characteristics of DevOps Teams

DevOps teams embrace both **cultural change** and **technology practices**:

#### Key Characteristics:
- Collaboration, trust, and transparency among team members
- Use of **Agile methodologies** for iterative development
- Faster and frequent delivery of small features
- **CI/CD automation** for code integration and deployment
- Automated testing and monitoring of applications
- Continuous improvement through feedback and performance metrics

````python
from IPython.display import Image
Image('./img/cicd3.png')
````

#### 🔄 Modern Software Delivery Lifecycle

Modern development follows a **cyclical workflow**:

1. **Plan** – gather requirements, define user stories, and plan sprint tasks  
2. **Develop** – write code  
3. **Build** – compile or package code  
4. **Test** – validate functionality through testing  
5. **Release & Deploy** – deliver the application to environments  
6. **Monitor & Improve** – track performance and enhance product  

This cycle repeats continuously for ongoing improvement.


#### 🔧 What is Continuous Integration (CI)?

With many developers contributing code, CI ensures:
- Code is continuously integrated into a shared repository (e.g., Git)
- Automated builds are triggered for every code merge
- Automated tests run with each build

#### 🎯 Goal of CI:
Catch errors early, improve code quality, and maintain a **stable build**.


#### 🚀 Continuous Delivery vs Continuous Deployment

Once code is tested, it must be released and deployed.

| Concept | Explanation |
|----------|----------------|
| **Continuous Delivery (CD)** | Code is automatically prepared for deployment. Manual approval is required before going to PROD. |
| **Continuous Deployment** | No manual intervention. Code is automatically deployed to production once tests pass. |

Both extend CI to ensure reliable and frequent releases.

**Continuous improvement** is achieved through observing the performance of the application and acting upon that.


#### 🧩 CI/CD in Azure Data Factory (ADF)

You might wonder:
> “Where is the code in ADF if we design with drag-and-drop?”

#### 📍 Code in ADF
Every ADF object is stored as **JSON** (pipelines, dataflows, datasets, triggers, etc.).  
This JSON is treated as **code**, enabling version control.

#### 🏗 Build in ADF
- In programming languages (e.g., Java), a build generates a deployable artifact (e.g., JAR file)
- For ADF, a build produces an **ARM Template**

#### 📦 What is an ARM Template?
- ARM = Azure Resource Manager  
- A JSON file that defines the ADF configuration and objects  
- Used to deploy the ADF project to other environments (Test, QA, Prod)



#### 🧪 Testing ADF Code

- Automated testing is common in development languages (e.g., unit tests)
- Automated testing for ADF is **not as straightforward**
- Many teams rely on **manual testing** for ADF
- A .NET testing framework exists for ADF (optional to explore)


#### 🚢 Release & Deployment

After testing:
- The ARM Template is **released and deployed** to higher environments
- Deployment creates ADF components (pipelines, triggers, datasets, dataflows, etc.) in target Data Factory

> There are some ADF-specific considerations for deployment (e.g., parameterization, environment variables), which will be covered later.

#### 📌 Git Integration for ADF (Git Mode)

To enable collaboration, versioning, and review process, we integrate ADF with Git.

#### Supported Git Repositories

- Azure DevOps Repos ✅ *(we will use this)*
- GitHub

#### How Git Mode Helps

| Capability | Benefit |
|-------------|------------|

| Commit changes | Save work incrementally |
| Pull Requests | Enable review & approvals |
| Version Control | Rollback changes |
| Merge to master/main | Controlled integration |

#### Flow with Git Mode

1. Developer creates **feature branch**
2. Makes changes and commits
3. Submits PR for review
4. Merge into **main/master** branch
5. Click **Publish** in ADF

---

#### 🧱 What Happens When We Click Publish?

The **Publish** button performs a **manual build** step.

On publish, two things happen:

| Output | Explanation |
|----------|----------------|
| Changes published to ADF Dev environment | Updates Data Factory service |
| **`adf_publish` branch generated** | Contains ARM templates for deployment |

#### What is in the `adf_publish` Branch?

The branch contains:

- ARM Templates (JSON files)
- Parameters file for environment variables

These files are used for deploying ADF to higher environments (Test, Prod) via Release Pipeline.

> 📍 **Publish = Build step of Continuous Integration**

---

#### 🚀 Deployment to Higher Environments

After Publish creates ARM templates, an **Azure DevOps Release Pipeline** deploys them to:

- Test
- Production

So, in **Option 1**:

| Stage | Automation |
|----------|----------------|
| Build (Publish) | ❌ Manual |
| Release / Deploy | ✅ Automated |

This is why Option 1 is called **Manual Build + Automated Deploy**.

---

#### 🧠 Summary of Option 1 – ADF Publish Method

- Developers work in Git mode with feature branches
- Manual **Publish** generates ARM templates (`adf_publish` branch)
- Azure DevOps Release Pipeline deploys ARM templates to Test/Prod
- CI is partially automated (manual build), CD is automated

This option is still used in many companies and interviews may refer to it as:

- **ADF Publish Approach**
- **Manual Build – Automated Deploy**
- **Option 1 CI/CD for ADF**

#### <b> 120. Configuring Git Integration with Azure Data Factory</b>
#### 📘 Overview
This notebook provides detailed notes on how to configure a Git repository for Azure Data Factory (ADF) development. It also explains key ADF Git concepts such as collaboration branches, feature branches, and the `adf_publish` branch.

#### 🔑 Key Git Concepts for ADF
Before configuring Git integration, it’s important to understand a few ADF‑specific Git terminologies:

#### ✔ Recommended Branching Strategy
Microsoft recommends the following branching approach for ADF:

| Branch | Purpose |
|--------|-----------|
| **Main/Master (Collaboration Branch)** | The only branch from which code can be **Published** to ADF. No direct development allowed. |
| **Feature Branches** | Created from Main for development work. Merged via Pull Request (PR). |
| **adf_publish** | Auto‑generated branch after Publish. Contains ARM templates for deployment. |

🔹 **Developers must never develop directly on main.**

Publishing from main triggers creation/update of the `adf_publish` branch containing ARM templates required for deployments.

````python
from IPython.display import Image
Image('./img/cicd11.png')
````

#### 🧰 Step 1: Create a Git Repository in Azure DevOps
You can either:
- Create a repo in Azure DevOps first **(recommended)**, then link ADF to it, or
- Let ADF create the repo for you

Here is the recommended approach:

1. Go to Azure DevOps > Select your Project
2. Navigate to **Repos**
3. Click **New Repository**
4. Name the repository same as the ADF instance for easy mapping

> ✅ Recommended Naming: Use the same name as the ADF instance (e.g., `dev-ci-cd-demo-adf`)

#### 🔧 Step 2: Configure Branch Policies
To enforce proper CI/CD workflows:

1. Go to **Repos → Branches**
2. Select the `main` branch
3. Click **... → Branch Policies**
4. Enable:
   - ✅ Minimum number of reviewers (e.g., 2 — but use 1 for learning/demo)
   - ✅ Require pull request to merge

📍 In real projects: *Do NOT allow requestors to approve their own PRs*

> This ensures quality code review and prevents direct commits to main.

#### 🔗 Step 3: Configure Git in Azure Data Factory
1. Open **ADF Studio** for the Dev Data Factory
2. Navigate to one of the following:
   - Home screen → **Set up Code Repository**
   - Author tab → Git dropdown → **Set up Code Repository**
   - Manage tab → **Git Configuration → Configure**

#### Fill in Configuration:
| Field | Value |
|-------|--------|
| Repository Type | Azure DevOps Git |
| Azure AD Tenant | Default Directory (or the one used for ADF resources) |
| DevOps Organization | Select your organization |
| Project | Select the DevOps project created |
| Repository | Select the repo created (e.g., `dev-ci-cd-demo-adf`) |
| Collaboration Branch | `main` or `master` |
| Publish Branch | `adf_publish` (auto‑selected) |

📍 If moving existing pipelines to Git, enable *Import Existing Resources*. For a clean start, leave unchecked.

#### 🧪 After Configuration
Once the setup is complete:

- ADF switches to **Git mode** instead of Live mode
- You will see branch name at the top bar of ADF Studio
- You must now **create a feature branch** before making changes

⚠️ If you try to work on main branch, ADF will show an error because branch policies prevent direct edits.
